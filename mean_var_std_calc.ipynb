```python
import numpy as np
```

```python
data_in = [0, 1, 2, 3, 4, 5, 6, 7, 8]
print('data input',data_in)
```

```python
if len(data_in) != 9:
  raise ValueError('list must contain nine numbers')
```

```python
matrix = np.array(data_in).reshape(3,3)
print('matrix',matrix)
```

```python
mean_axis0 = matrix.mean(axis=0).tolist()
mean_axis1 = matrix.mean(axis=1).tolist()
mean_flat  = matrix.mean().tolist()

print('Mean Axis 0:',mean_axis0)
print('Mean Axis 1:',mean_axis1)
print('Mean Flat:',mean_flat)
```

```python
def calculate(list_input):
  if len(list_input) != 9:
    raise ValueError('List must contain nine numbers.')

  matrix = np.array(list_input).reshape(3,3)

  calculations = {
      'mean': [
          matrix.mean(axis=0).tolist(),
          matrix.mean(axis=1).tolist(),
          matrix.mean().tolist()
      ],
      'variance': [
          matrix.var(axis=0).tolist(),
          matrix.var(axis=1).tolist(),
          matrix.var().tolist()
      ],
      'standard deviation':[
          matrix.std(axis=0).tolist(),
          matrix.std(axis=1).tolist(),
          matrix.std().tolist()
      ],
      'max':[
          matrix.max(axis=0).tolist(),
          matrix.max(axis=1).tolist(),
          matrix.max().tolist()
      ],
      'min':[
          matrix.min(axis=0).tolist(),
          matrix.min(axis=1).tolist(),
          matrix.min().tolist()
      ],
      'sum':[
          matrix.sum(axis=0).tolist(),
          matrix.sum(axis=1).tolist(),
          matrix.sum().tolist()
      ]
  }

  return calculations
```

```python
calculate([0, 1, 2, 3, 4, 5, 6, 7, 8])
```
